### Pre-processing: drop unnecessary columns and split the ASV hashes into a separate column

```python
dwc = dwc.drop(['datasetID', 
                'basisOfRecord', 
                'recordedBy', 
                'parentEventID',
                'organismQuantityType', 
                'occurrenceStatus', 
                'associatedSequences', 
                'occurrenceRemarks', 
                'sampleSizeUnit'], axis=1)
dwc[['ASV', 'hash']] = dwc['taxonID'].str.split(':', expand=True)
dwc.head()
```

### Pre-processing: Separate the table into 16S and 18S data sets

```python
dwc[['parentEventID', 'markerGene']] = dwc['eventID'].str.rsplit('_', n=1, expand=True)
dwc.head()
```

```python
gomecc_16S = dwc[dwc['markerGene'] == '16S']
gomecc_18S = dwc[dwc['markerGene'] == '18S']
gomecc_16S.head()
```

### 1. Make the ASV tables

```python
gomecc_16S_asvs = gomecc_16S[['parentEventID', 'organismQuantity', 'hash']]
gomecc_18S_asvs = gomecc_18S[['parentEventID', 'organismQuantity', 'hash']]
gomecc_16S_asvs.head()
gomecc_18S_asvs.head()
```

#### Convert to wide format

```python
gomecc_16S_asvs_wide = gomecc_16S_asvs.groupby(['parentEventID', 'hash']).organismQuantity.first().unstack().fillna(0).T
gomecc_18S_asvs_wide = gomecc_18S_asvs.groupby(['parentEventID', 'hash']).organismQuantity.first().unstack().fillna(0).T
```

#### Save as csv files

```python
gomecc_16S_asvs_wide.to_csv('GOMECC_16S_ASV_table.csv')
gomecc_18S_asvs_wide.to_csv('GOMECC_18S_ASV_table.csv')
```

### 2. Make the taxonomy tables

```python
def format_taxonomy_table(df, gene):
    tax = df[['verbatimIdentification', 'taxonID', 'kingdom', 'phylum', 'class', 'order', 'family', 'genus']]
    tax.loc[:, 'hash'] = tax.loc[:, 'taxonID'].str.split(':', expand=True)[1]
    tax.loc[:, ['Domain', 'Phylum', 'Class', 'Order', 'Family', 'Genus', 'Species']] = tax.loc[:, 'verbatimIdentification'].str.split(';', n=6, expand=True)
    tax = tax.drop(['verbatimIdentification', 'taxonID'], axis=1)
    # Some 18S 'verbatimIdentification' fields only have species name; for these we pull out the WoRMS taxonomy fields
    if (gene=='18S'):
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Bacteria') & (tax['Domain'] != 'Archaea'), 'Species'] = tax['Domain']
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Bacteria') & (tax['Domain'] != 'Archaea'), 'Class'] = tax['class']
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Bacteria') & (tax['Domain'] != 'Archaea'), 'Order'] = tax['order']
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Bacteria') & (tax['Domain'] != 'Archaea'), 'Family'] = tax['family']
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Bacteria') & (tax['Domain'] != 'Archaea'), 'Genus'] = tax['genus']
        # Some unassigned 18S taxa have "nan" as Domain, we can rename these to be 'Unassigned'
        tax.loc[(tax['Species'] == 'Unassigned'), 'Domain'] = 'Unassigned'
        # The WoRMS "kingdoms" are all eukaryotic (fungi, protozoa, chromista, etc.) so we assign all those to 'Eukaryota' 
        tax.loc[(tax['Domain'] != 'Eukaryota') & (tax['Domain'] != 'Unassigned'), 'Domain'] = 'Eukaryota'
        pass
    tax = tax.drop(['kingdom', 'phylum', 'class', 'order', 'family', 'genus'], axis=1)
    tax = tax.drop_duplicates()
    return(tax)
```

```python
tax_16S = format_taxonomy_table(gomecc_16S, '16S')
tax_18S = format_taxonomy_table(gomecc_18S, '18S')
```

```python
tax_18S.head()
```

```python
tax_16S.to_csv('GOMECC_16S_taxonomy_table.csv', index=None)
tax_18S.to_csv('GOMECC_18S_taxonomy_table.csv', index=None)
```

#### 16S

```python
gomecc_16S_metadata = gomecc_16S[['parentEventID', 'materialSampleID', 'eventID', 'eventDate', 'samplingProtocol', 'locationID',
                                 'waterBody', 'locality', 'minimumDepthInMeters', 'decimalLatitude', 'decimalLongitude']]
# Need to fill in some missing metadata!
gomecc_16S_metadata['waterBody'] = gomecc_16S_metadata['waterBody'].fillna('Mexico, Gulf of')
gomecc_16S_metadata['# ASVs'] = 1
gomecc_16S_metadata_grouped = gomecc_16S_metadata.groupby(['parentEventID', 'materialSampleID', 'eventID', 'eventDate', 'samplingProtocol', 'locationID',
                                 'waterBody', 'locality', 'minimumDepthInMeters', 'decimalLatitude', 'decimalLongitude']).sum().reset_index()
```

#### 18S

```python
gomecc_18S_metadata = gomecc_18S[['parentEventID', 'materialSampleID', 'eventID', 'eventDate', 'samplingProtocol', 'locationID',
                                 'waterBody', 'locality', 'minimumDepthInMeters', 'decimalLatitude', 'decimalLongitude']]
# Need to fill in some missing metadata!
gomecc_18S_metadata['waterBody'] = gomecc_18S_metadata['waterBody'].fillna('Mexico, Gulf of')
gomecc_18S_metadata['# ASVs'] = 1
gomecc_18S_metadata_grouped = gomecc_18S_metadata.groupby(['parentEventID', 'materialSampleID', 'eventID', 'eventDate', 'samplingProtocol', 'locationID',
                                 'waterBody', 'locality', 'minimumDepthInMeters', 'decimalLatitude', 'decimalLongitude']).sum().reset_index()
```
